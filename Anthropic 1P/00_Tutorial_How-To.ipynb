## Usage Notes & Tips 💡

All prompt engineering techniques in this course still apply.

### The `claude` CLI
Every notebook in this tutorial calls `get_completion()` from [claude_cli.py](claude_cli.py), a small wrapper that runs `claude -p "<prompt>" --model ... --tools "" --system-prompt ...` and returns the printed response. See that file's docstrings for exactly how it maps the original `prompt` / `system_prompt` / `prefill` (and, in the appendices, `messages` / `stop_sequences`) arguments onto CLI flags, and where the mapping is an approximation.

Below is an example of what running a prompt will look like in this tutorial. Run the next cell to import `get_completion`.

```python
from claude_cli import get_completion
```

Feel free to play around with the prompt string to elicit different responses from Claude.

```python
# Prompt
prompt = "Hello, Claude!"

# Get Claude's response
print(get_completion(prompt))
```
